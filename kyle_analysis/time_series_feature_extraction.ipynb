```python
import json, time, pandas as pd

QUERY_COUNT = 0x2
SUM_FILE_SIZES = 0x40
PARTS_MASK = QUERY_COUNT | SUM_FILE_SIZES  # 0x42

def _wait_bundle(syn, table_id, job_id, sleep_s=1.0, timeout_s=300):
    start = time.time()
    while True:
        res = syn.restGET(f"/entity/{table_id}/table/query/async/get/{job_id}")
        if isinstance(res, dict) and res.get("jobState") in ("PROCESSING", "PENDING"):
            if time.time() - start > timeout_s:
                raise TimeoutError(f"Timed out waiting for query job {job_id}")
            time.sleep(sleep_s)
            continue
        return res

def table_total_file_bytes(syn, table_id: str):
    sql = f"SELECT * FROM {table_id}"
    body = {"query": {"sql": sql}, "partMask": PARTS_MASK}

    # IMPORTANT: send JSON string, not a raw dict
    job = syn.restPOST(
        f"/entity/{table_id}/table/query/async/start",
        json.dumps(body),
        headers={"Content-Type": "application/json"}
    )

    job_id = job.get("token") or job.get("jobId")
    bundle = _wait_bundle(syn, table_id, job_id)

    return {
        "table_id": table_id,
        "rowCount": bundle.get("queryCount"),
        "sumFileSizesBytes": bundle.get("sumFileSizesBytes"),
    }

def bytes_to_gb(n):
    return None if n is None else n / (1024**3)

tables = ["syn5511449", "syn5511444", "syn5511439", "syn5511434"]
rows = []
for t in tables:
    stats = table_total_file_bytes(syn, t)
    stats["sumFileSizesGB"] = bytes_to_gb(stats["sumFileSizesBytes"])
    rows.append(stats)

pd.DataFrame(rows).sort_values("sumFileSizesBytes", ascending=False)
```

## Helper functions


```python
def is_valid_audio(file_path):
    try:
        audio = AudioSegment.from_file(file_path, format="m4a")
        return len(audio) > 100
    except Exception as e:
        dprint(f"is_valid_audio: could not read {file_path}: {e}")
        return False


def convert_to_wav(input_file, output_dir, output_name=None):
    """
    Convert `input_file` (m4a) to a .wav file inside `output_dir`.
    """
    base_name = output_name if output_name is not None else os.path.splitext(os.path.basename(input_file))[0]
    output_path = os.path.join(output_dir, f"{base_name}.wav")

    if os.path.exists(output_path):
        dprint(f"convert_to_wav: {output_path} already exists, skipping re-conversion")
        return output_path

    audio = None
    for fmt in ("m4a", "mp4"):
        try:
            audio = AudioSegment.from_file(input_file, format=fmt)
            break
        except Exception as e:
            dprint(f"convert_to_wav: decode as {fmt} failed for {input_file}: {e}")

    if audio is None:
        print(f"[ERROR] convert_to_wav: could not decode {input_file} as m4a or mp4")
        return None

    if len(audio) <= 100:
        dprint(f"convert_to_wav: skipping invalid or empty file (<=100ms): {input_file}")
        return None

    try:
        audio.export(output_path, format="wav")
        dprint(f"convert_to_wav: wrote {output_path}")
        return output_path
    except Exception as e:
        print(f"[ERROR] convert_to_wav: export failed for {input_file}: {e}")
        return None


def convert_folder_to_wav(input_dir):
    m4a_files = [f for f in os.listdir(input_dir) if f.endswith(".m4a")]
    print(f"[convert_folder_to_wav] {len(m4a_files)} .m4a files in {input_dir}")
    for filename in tqdm(m4a_files, desc="convert_folder_to_wav"):
        input_path = os.path.join(input_dir, filename)
        health_code = os.path.splitext(filename)[0]
        wav_path = os.path.join(input_dir, f"{health_code}.wav")
        if os.path.exists(wav_path):
            dprint(f"convert_folder_to_wav: {wav_path} already exists, skipping")
            continue
        convert_to_wav(input_path, input_dir)


# Adapted from https://github.com/Sage-Bionetworks/mPower-sdata/blob/master/examples/mPower-bootstrap.py
# Downloads ALL matching audio files, customize batch size accordingly.
def extract_audio_files(output_dir, diagnosis=False, batch_size=100, limit=None):
    """
    `limit`, if set, caps the number of healthCodes fetched -- e.g.
    extract_audio_files(..., limit=50) downloads+converts just 50 real
    recordings via the exact same Synapse query path as a full run, instead
    of the whole dataset. Used by fetch_sample_df_from_synapse below to get
    a small batch of genuinely real healthCode values without running the
    full extraction.
    """
    os.makedirs(output_dir, exist_ok=True)

    # Load and filter survey data
    survey_data = pd.read_csv(CSV_FILES_PATH + "survey_data.csv")
    survey_data = survey_data[survey_data["professional-diagnosis"] == diagnosis]
    all_healthcodes = survey_data["healthCode"].tolist()

    if limit is not None:
        all_healthcodes = all_healthcodes[:limit]

    print(f"[extract_audio_files] {len(all_healthcodes)} healthCodes to fetch "
          f"(diagnosis={diagnosis}, limit={limit}) -> {output_dir}")

    all_mappings = []
    start = time.time()

    for batch_num, i in enumerate(range(0, len(all_healthcodes), batch_size)):
        batch = all_healthcodes[i:i + batch_size]
        healthcodes_str = "','".join(batch)

        batch_start = time.time()
        query = f"SELECT * FROM syn5511444 WHERE healthCode IN ('{healthcodes_str}')"
        subset_query = syn.tableQuery(query)
        subset_df = subset_query.asDataFrame()
        subset_df["audio_audio.m4a"] = subset_df["audio_audio.m4a"].astype(str)
        file_map = syn.downloadTableColumns(subset_query, "audio_audio.m4a")
        dprint(f"[batch {batch_num}] queried {len(subset_df)} rows, "
               f"downloaded {len(file_map)} files in {time.time() - batch_start:.1f}s")

        for file_handle_id, m4a_path in file_map.items():
            matched_row = subset_df[subset_df["audio_audio.m4a"] == file_handle_id]
            if matched_row.empty:
                print(f"[WARN] file handle ID {file_handle_id} not found in subset_df, skipping")
                continue

            healthcode = matched_row.iloc[0]["healthCode"]
            record_id = matched_row.iloc[0]["recordId"]

            # FIX: use recordId (stable, unique per row) as the output
            # filename instead of Synapse's randomized local cache filename.
            wav_path = convert_to_wav(m4a_path, output_dir, output_name=record_id)

            if wav_path is None:
                print(f"[WARN] no wav produced for recordId={record_id} (healthCode={healthcode})")
                continue

            all_mappings.append({
                "record_id": record_id,
                "healthCode": healthcode,
                "wav_path": wav_path,
            })

        elapsed = time.time() - start
        n_done = i + len(batch)

        # Provide ETA estimate
        rate = elapsed / max(1, n_done)
        remaining = rate * (len(all_healthcodes) - n_done)
        print(f"[extract_audio_files] batch {batch_num}: {n_done}/{len(all_healthcodes)} "
              f"healthCodes processed ({100 * n_done / max(1, len(all_healthcodes)):.1f}%), "
              f"{len(all_mappings)} wavs so far, {elapsed/60:.1f} min elapsed, "
              f"~{remaining/60:.1f} min remaining (ETA)")

    print(f"[extract_audio_files] done: {len(all_mappings)} wavs in {(time.time() - start)/60:.1f} min")
    return pd.DataFrame(all_mappings)


def extract_time_series_features_for_file(wav_path, record_id, health_code, gender_flag,
                                            diagnosis=None,
                                            time_step=PITCH_TIME_STEP, window_size=JITTER_WINDOW_SIZE,
                                            verbose=True):
    """
    Run the full per-file feature extraction (F0, F1-F3, jitter, shimmer, HNR
    over time) for a single wav file. Returns a one-row-per-frame DataFrame,
    or None if the file couldn't be processed.

    Factored out of get_formants_over_time_parquet so the exact same logic
    can be reused by both the full pipeline and the small sample/baseline
    run below.

    `time_step` and `window_size` default to PITCH_TIME_STEP /
    JITTER_WINDOW_SIZE (see "Extraction hyperparameters" above) but can be
    overridden per-call if you want to experiment without touching the
    globals.

    `diagnosis` ("positive"/"negative") is stamped onto every output row.
    Now that positive and negative recordings are saved into a single
    combined parquet file (see Config), this is what lets you filter back
    out by diagnosis later -- without it there'd be no way to tell the two
    groups apart once merged.
    """
    t0 = time.time()
    if verbose:
        dprint(f"[{record_id}] loading {wav_path} (time_step={time_step}s, window_size={window_size}s)")

    if health_code is None or (isinstance(health_code, float) and pd.isna(health_code)):
        print(f"[WARN] [{record_id}] missing healthCode -- this row won't be joinable by "
              f"healthCode later (recordId is still preserved)")

    if not os.path.exists(wav_path):
        print(f"[ERROR] [{record_id}] file does not exist: {wav_path}")
        return None

    max_formant = 5500 if gender_flag == 1 else 5000

    try:
        audio, sr = librosa.load(wav_path, sr=None)
    except Exception as e:
        print(f"[ERROR] [{record_id}] librosa.load failed for {wav_path}: {e}")
        return None

    try:
        trimmed_audio, idx = clip_audio(audio)
        sound = parselmouth.Sound(trimmed_audio, sampling_frequency=sr)
    except Exception as e:
        print(f"[ERROR] [{record_id}] clip_audio/Sound construction failed: {e}")
        return None

    if verbose:
        dprint(f"[{record_id}] loaded + trimmed, duration={sound.get_total_duration():.2f}s "
               f"(t+{time.time() - t0:.2f}s)")

    try:
        pitch = sound.to_pitch(time_step)
        n_frames = pitch.get_number_of_frames()
        f0_times = [pitch.get_time_from_frame_number(i) for i in range(1, n_frames + 1)]
        f0_values = [pitch.get_value_in_frame(i) for i in range(1, n_frames + 1)]

        formants = sound.to_formant_burg(time_step, 5, max_formant, 0.025, 50)
    except Exception as e:
        print(f"[ERROR] [{record_id}] pitch/formant extraction failed: {e}")
        return None

    if verbose:
        dprint(f"[{record_id}] pitch + formants computed, {n_frames} frames (t+{time.time() - t0:.2f}s)")

    f1, f2, f3 = [], [], []
    jitter_values, shimmer_values, hnr_values = [], [], []

    log_every = max(1, n_frames // 5)  # ~5 progress lines per file when verbose
    for frame_idx, t in enumerate(f0_times):
        f1.append(formants.get_value_at_time(1, t))
        f2.append(formants.get_value_at_time(2, t))
        f3.append(formants.get_value_at_time(3, t))

        try:
            start_time = max(0, t - window_size / 2)
            end_time = min(sound.get_total_duration(), t + window_size / 2)

            window_sound = sound.extract_part(
                from_time=start_time,
                to_time=end_time,
                preserve_times=False
            )

            localJitter, _, _, _, _ = get_jitter(window_sound, minF0=75, maxF0=500)
            localShimmer, _, _, _, _ = get_shimmer(window_sound, minF0=75, maxF0=500)
            hnr = get_harmonic_to_noise_ratio(window_sound, minF0=75)

        except Exception as e:
            if verbose:
                dprint(f"[{record_id}] frame {frame_idx}/{n_frames} @ t={t:.2f}s: "
                       f"jitter/shimmer/hnr failed ({e})")
            localJitter = np.nan
            localShimmer = np.nan
            hnr = np.nan

        jitter_values.append(localJitter)
        shimmer_values.append(localShimmer)
        hnr_values.append(hnr)

        if verbose and frame_idx % log_every == 0:
            dprint(f"[{record_id}] frame {frame_idx}/{n_frames} (t+{time.time() - t0:.2f}s)")

    # recordId (per-recording), healthCode (per-person), and diagnosis are
    # all included on every row, front and center: recordId/healthCode so
    # this time-series data can be joined against future datasets at either
    # the recording or person level, and diagnosis so positive/negative can
    # still be filtered back apart even though they're saved to one file.
    time_df = pd.DataFrame({
        "recordId": record_id,
        "healthCode": health_code,
        "diagnosis": diagnosis,
        "time": f0_times,
        "F0": f0_values,
        "F1": f1,
        "F2": f2,
        "F3": f3,
        "Jitter (local)": jitter_values,
        "Shimmer (local)": shimmer_values,
        "Harmonic to Noise Ratio": hnr_values,
    })

    if verbose:
        dprint(f"[{record_id}] done: {len(time_df)} rows in {time.time() - t0:.2f}s")

    return time_df


def get_formants_over_time_parquet(df, output_parquet_path,
                                    time_step=PITCH_TIME_STEP, window_size=JITTER_WINDOW_SIZE,
                                    checkpoint_every=200, verbose_per_file=False):
    all_time_series = []
    n_total = len(df)
    n_ok, n_missing, n_failed = 0, 0, 0
    start = time.time()

    print(f"[get_formants_over_time_parquet] starting on {n_total} rows -> {output_parquet_path}")
    print(f"[get_formants_over_time_parquet] time_step={time_step}s, window_size={window_size}s "
          f"(~{1/time_step:.0f} frames/sec of audio, see 'Extraction hyperparameters' above for what these do)")
    if "diagnosis" in df.columns:
        dprint(f"[get_formants_over_time_parquet] diagnosis breakdown: "
               f"{df['diagnosis'].value_counts().to_dict()}")

    n_missing_healthcode = 0

    for row in tqdm(df.itertuples(index=False), total=n_total, desc=os.path.basename(output_parquet_path)):
        wav = getattr(row, "wav_path", None)
        record_id = getattr(row, "record_id", None)
        health_code = getattr(row, "healthCode", None)
        diagnosis = getattr(row, "diagnosis", None)
        gender_str = str(getattr(row, "gender", "unknown")).lower()
        gender_flag = 1 if gender_str == "female" else 0

        if health_code is None or (isinstance(health_code, float) and pd.isna(health_code)):
            n_missing_healthcode += 1

        if not wav or not os.path.exists(wav):
            print(f"[WARN] [{record_id}] missing wav, skipping: {wav}")
            n_missing += 1
            continue

        time_df = extract_time_series_features_for_file(
            wav, record_id, health_code, gender_flag, diagnosis=diagnosis,
            time_step=time_step, window_size=window_size, verbose=verbose_per_file
        )

        if time_df is None:
            n_failed += 1
            continue

        all_time_series.append(time_df)
        n_ok += 1

        if checkpoint_every and n_ok % checkpoint_every == 0:
            checkpoint_df = pd.concat(all_time_series, ignore_index=True)
            checkpoint_df.to_parquet(output_parquet_path, index=False)
            dprint(f"[checkpoint] saved {n_ok} files ({len(checkpoint_df)} rows) "
                   f"to {output_parquet_path} at {time.time() - start:.1f}s")

    elapsed = time.time() - start
    print(f"[get_formants_over_time_parquet] done in {elapsed:.1f}s: "
          f"{n_ok} ok, {n_missing} missing, {n_failed} failed (of {n_total})")
    if n_missing_healthcode:
        print(f"[WARN] {n_missing_healthcode}/{n_total} rows had no healthCode -- "
              f"those rows' time-series data is still saved (recordId is preserved) "
              f"but won't be joinable by healthCode later")

    if all_time_series:
        full_df = pd.concat(all_time_series, ignore_index=True)
        full_df.to_parquet(output_parquet_path, index=False)
        print(f"[OK] saved all data to: {output_parquet_path}")
        return full_df
    else:
        print("[WARN] no valid data to save.")
        return pd.DataFrame()
```

```python
from concurrent.futures import ProcessPoolExecutor, as_completed

def _extract_one(args):
    wav, record_id, health_code, gender_flag, diagnosis, time_step, window_size = args
    return extract_time_series_features_for_file(
        wav, record_id, health_code, gender_flag, diagnosis=diagnosis,
        time_step=time_step, window_size=window_size, verbose=False
    )

# PARALELLIZED VERSION OF GET FORMANTS OVER TIME
def get_formants_over_time_parquet_parallel(df, output_parquet_path,
                                             time_step=PITCH_TIME_STEP, window_size=JITTER_WINDOW_SIZE,
                                             n_workers=N_WORKERS, checkpoint_every=200,
                                             heartbeat_every=50, heartbeat_seconds=60):
    """
    Same as get_formants_over_time_parquet, but processes `n_workers` files
    at once in separate processes instead of one file at a time.

    FIX (debugging): in addition to the tqdm progress bar, this prints a
    plain heartbeat line (files done, %, elapsed, ETA) every `heartbeat_every`
    completions or `heartbeat_seconds` seconds, whichever comes first. This
    is deliberately a plain print(), not tqdm -- tqdm's progress bar updates
    a single line in place using carriage returns, which often doesn't
    render as readable output if you're not watching the notebook live (a
    log file, a terminal that doesn't handle \\r well, etc.) -- exactly the
    situation where a run that's actually progressing normally can look
    "hung forever" with no visible evidence otherwise.
    """
    tasks = [
        (
            row.wav_path,
            row.record_id,
            row.healthCode,
            1 if str(getattr(row, "gender", "unknown")).lower() == "female" else 0,
            getattr(row, "diagnosis", None),
            time_step,
            window_size,
        )
        for row in df.itertuples(index=False)
        if isinstance(row.wav_path, str) and os.path.exists(row.wav_path)
    ]
    n_total = len(tasks)
    print(f"[parallel] {n_total}/{len(df)} rows have an existing wav file, "
          f"running with {n_workers} workers (time_step={time_step}s, window_size={window_size}s, "
          f"mp_context={'fork' if _MP_CONTEXT else 'default'})")

    if SAMPLE_BASELINE_AVG_SEC_PER_FILE is not None and n_total:
        projected = SAMPLE_BASELINE_AVG_SEC_PER_FILE * n_total / max(1, n_workers)
        print(f"[parallel] projected time, based on the sample baseline's "
              f"{SAMPLE_BASELINE_AVG_SEC_PER_FILE:.2f}s/file and {n_workers} workers: "
              f"~{projected/60:.0f} min (~{projected/3600:.1f} hr) -- rough estimate, "
              f"actual audio lengths vary")
    else:
        print(f"[parallel] no pre-run estimate yet -- you do NOT need to run the sample "
              f"baseline first. A real ETA will print automatically below after the first "
              f"{heartbeat_every} files finish (or {heartbeat_seconds}s, whichever comes "
              f"first), calibrated from this run's own timing")

    all_time_series = []
    n_ok, n_failed = 0, 0
    start = time.time()
    last_heartbeat_time = start
    last_heartbeat_n = 0

    with ProcessPoolExecutor(max_workers=n_workers, mp_context=_MP_CONTEXT) as pool:
        futures = {pool.submit(_extract_one, t): t for t in tasks}
        for fut in tqdm(as_completed(futures), total=len(futures), desc="parallel extraction"):
            wav, record_id = futures[fut][0], futures[fut][1]
            try:
                time_df = fut.result()
            except Exception as e:
                print(f"[ERROR] [{record_id}] worker raised: {e}")
                time_df = None

            if time_df is None:
                n_failed += 1
            else:
                all_time_series.append(time_df)
                n_ok += 1

            n_done = n_ok + n_failed
            now = time.time()
            if (n_done - last_heartbeat_n >= heartbeat_every) or (now - last_heartbeat_time >= heartbeat_seconds):
                elapsed = now - start
                rate = elapsed / max(1, n_done)
                remaining = rate * (n_total - n_done)
                print(f"[parallel] heartbeat: {n_done}/{n_total} done "
                      f"({100 * n_done / max(1, n_total):.1f}%), {n_ok} ok, {n_failed} failed, "
                      f"{elapsed/60:.1f} min elapsed, ~{remaining/60:.1f} min remaining (ETA)")
                last_heartbeat_time = now
                last_heartbeat_n = n_done

            if checkpoint_every and n_ok and n_ok % checkpoint_every == 0:
                pd.concat(all_time_series, ignore_index=True).to_parquet(output_parquet_path, index=False)
                dprint(f"[checkpoint] saved {n_ok} files at {time.time() - start:.1f}s")

    elapsed = time.time() - start
    print(f"[parallel] done in {elapsed/60:.1f} min: {n_ok} ok, {n_failed} failed (of {n_total})")

    if all_time_series:
        full_df = pd.concat(all_time_series, ignore_index=True)
        full_df.to_parquet(output_parquet_path, index=False)
        print(f"[OK] saved all data to: {output_parquet_path}")
        return full_df
    return pd.DataFrame()
```

## Stabilization metrics (sample baseline only, for now)

Ported over from `stabilizationmetrics.ipynb`, adapted to run directly on
`sample_result` (the sample baseline, above) rather than a full
positive/negative parquet. Scoped to the sample baseline only, as agreed --
running this against the full dataset is future work once the full
extraction itself has been run.

**Two things fixed while porting:**

1. `compute_tts_dataframe` accepted a `diffs` parameter but the actual call
   to `compute_time_to_stabilization` inside it hardcoded `diffs=4`
   regardless of what was passed in -- so the commented-out "STRICT RUN:
   Window size 200ms" (`diffs=200`) in the original notebook would have had
   *no effect at all*. Fixed here to actually use the passed `diffs`.
2. `compute_slope_to_stabilization` / `compute_area_of_stability` are
   rewritten as explicit per-row loops instead of
   `groupby(...).apply(lambda g: pd.Series(...))`, which avoids a pandas
   `DeprecationWarning` that varies by pandas version, and is easier to
   follow. `np.trapz` (deprecated in recent NumPy) is replaced with
   `np.trapezoid` where available.

**Two output tables**, matching "can separate into different dataframes":

- **`stabilization_tts`** -- one row per (healthCode, recordId, feature) for
  feature in F0/F1/F2/F3: `t0_time`, `ts_time`, `time_to_stabilization_s`,
  `slope_to_stabilization`, `area_of_stability`. This is the direct
  equivalent of `tts_pos_full`/`tts_neg_full` in the original notebook.
- **`stabilization_means`** -- one row per (recordId, condition) for
  condition in {"before", "after", "across"}: `recordId`, `healthCode`,
  `tts`, `condition`, then the mean of F0, F1, F2, F3, Jitter (local),
  Shimmer (local), Harmonic to Noise Ratio within that window. Per your
  call, the single per-recording split point (`tts`) is **F0's**
  stabilization time -- pitch stabilizing is the standard definition of
  "phonation has stabilized" -- not a separate split per feature. "before"
  is `[t0_time, ts_time)`, "after" is `[ts_time, end of recording]`, and
  "across" is every frame in the recording regardless of stabilization. If
  F0 never stabilized for a recording, "before"/"after" are NaN but
  "across" is still computed.

Both are saved as `.parquet` files in a `stabilization_metrics/` folder next
to this notebook (not the USB drive's `parquet_files/` directory), since you
asked to keep this separate for now.


```python
def compute_time_to_stabilization(time_s, f, beta, diffs, eps=1e-8):
    """
    Parameters
    ----------
    time_s : array
        Times in seconds ex. [0.02, 0.04, ..., 11.00]
    f : array
        Feature values (e.g., F0, F1, F2, F3).

    Hyperparameters
    ----------
    beta : float
        Relative error tolerance for stability windows
    diffs : int
        Number of consecutive step-to-step differences required to form the stability window
        (default value uses 4 from paper).
    eps : float
        Small constant to avoid dividing by zero if f[k+i] is 0.

    Return Values
    -------
    t0_time : float
        Time of phonation onset (first non-NaN value in time series).
    ts_time : float
        Time at stabilization index k (earliest time at index k satisfying stabilization condition).
    Ts : float
        Time-to-stabilization in seconds (ts_time - t0_time).
        If stabilization not found, ts_time and Ts are NaN.
    """
    time_s = np.asarray(time_s, dtype=float)
    f = np.asarray(f, dtype=float)

    # EDGE CASE: Need at least (diffs + 1) valid points to check `diffs` consecutive diffs
    valid = ~np.isnan(f)
    if valid.sum() < (diffs + 1):
        return (np.nan, np.nan, np.nan)

    # Phonation onset t0: first non-NaN sample
    t0_idx = np.where(valid)[0][0]
    t0_time = float(time_s[t0_idx])

    # Slide the window to find the earliest stabilization index k
    last_start = len(f) - diffs - 1
    for k in range(t0_idx, last_start + 1):
        window_idx = np.arange(k, k + diffs + 1)
        if not valid[window_idx].all():
            continue

        ok = True
        for i in range(diffs):
            a = f[k + i]
            b = f[k + i + 1]
            if np.abs(b - a) > beta * max(np.abs(a), np.abs(b), eps):
                ok = False
                break

        if ok:
            ts_idx = k + diffs
            ts_time = float(time_s[ts_idx])
            Ts = ts_time - t0_time
            return (t0_time, ts_time, Ts)

    return (t0_time, np.nan, np.nan)


TTS_KEYS = ["healthCode", "recordId", "feature"]


def compute_tts_dataframe(df, features=("F0", "F1", "F2", "F3"), beta=0.05, diffs=4, verbose=False):
    """
    Compute time-to-stabilization for each (healthCode, recordId) group and
    each feature column. Returns a DataFrame with one row per (healthCode,
    recordId, feature).

    FIX: previously `diffs` was accepted as a parameter but silently
    ignored -- the internal call to compute_time_to_stabilization always
    hardcoded diffs=4. Fixed to actually thread `diffs` through.
    """
    results = []
    print(f"[stabilization_tts] computing for {df['recordId'].nunique()} recordings "
          f"x {len(features)} features (beta={beta}, diffs={diffs})")

    for (hc, rid), g in df.groupby(["healthCode", "recordId"], sort=False):
        g = g.sort_values("time")
        time_s = g["time"].to_numpy()

        for feat in features:
            f = g[feat].to_numpy()
            t0_time, ts_time, Ts = compute_time_to_stabilization(
                time_s=time_s, f=f, beta=beta, diffs=diffs
            )
            if verbose:
                dprint(f"[stabilization_tts] {rid} {feat}: t0={t0_time}, ts={ts_time}, Ts={Ts}")
            results.append({
                "healthCode": hc,
                "recordId": rid,
                "feature": feat,
                "t0_time": t0_time,
                "ts_time": ts_time,
                "time_to_stabilization_s": Ts,
            })

    return pd.DataFrame(results)


def compute_slope_to_stabilization(tts_df, time_series_df):
    """
    Adds slope_to_stabilization: (value at ts_time - value at t0_time) /
    (ts_time - t0_time), per (healthCode, recordId, feature).

    FIX: rewritten as an explicit loop over tts_df rows instead of
    groupby(...).apply(lambda g: pd.Series(...)) -- the original triggered a
    pandas DeprecationWarning about grouping columns, and behaves
    inconsistently across pandas versions. Same result, just computed
    row-by-row instead.
    """
    long_df = time_series_df.melt(
        id_vars=["healthCode", "recordId", "time"],
        value_vars=["F0", "F1", "F2", "F3"],
        var_name="feature",
        value_name="value",
    )

    slopes = []
    for _, row in tts_df.iterrows():
        hc, rid, feat = row["healthCode"], row["recordId"], row["feature"]
        t0, ts = row["t0_time"], row["ts_time"]

        if pd.isna(ts) or pd.isna(t0):
            slopes.append(np.nan)
            continue

        denom = ts - t0
        if denom == 0:
            slopes.append(np.nan)
            continue

        g = long_df[
            (long_df["healthCode"] == hc) & (long_df["recordId"] == rid) & (long_df["feature"] == feat)
        ].dropna(subset=["value"])
        if g.empty:
            slopes.append(np.nan)
            continue

        # nearest samples to t0 and ts (within the whole series)
        i0 = (g["time"] - t0).abs().idxmin()
        is_ = (g["time"] - ts).abs().idxmin()
        f0 = g.loc[i0, "value"]
        fs = g.loc[is_, "value"]
        slopes.append((fs - f0) / denom)

    out = tts_df.copy()
    out["slope_to_stabilization"] = slopes
    return out


def compute_area_of_stability(tts_df, time_series_df):
    """
    Adds area_of_stability: trapezoidal integral of the feature's values
    over [t0_time, ts_time], per (healthCode, recordId, feature).

    FIX: rewritten as an explicit loop (see compute_slope_to_stabilization
    for why), and uses np.trapezoid where available -- np.trapz is
    deprecated as of NumPy 2.0.
    """
    # NOTE: can't use getattr(np, "trapezoid", np.trapz) here -- Python
    # evaluates default arguments eagerly, so np.trapz would be looked up
    # (and raise AttributeError) even on NumPy versions where it was
    # removed entirely, regardless of whether "trapezoid" exists.
    trapezoid_fn = np.trapezoid if hasattr(np, "trapezoid") else np.trapz

    long_df = time_series_df.melt(
        id_vars=["healthCode", "recordId", "time"],
        value_vars=["F0", "F1", "F2", "F3"],
        var_name="feature",
        value_name="value",
    )

    areas = []
    for _, row in tts_df.iterrows():
        hc, rid, feat = row["healthCode"], row["recordId"], row["feature"]
        t0, ts = row["t0_time"], row["ts_time"]

        if pd.isna(ts) or pd.isna(t0):
            areas.append(np.nan)
            continue

        g = long_df[
            (long_df["healthCode"] == hc) & (long_df["recordId"] == rid) & (long_df["feature"] == feat)
        ].dropna(subset=["value"])
        g = g[(g["time"] >= t0) & (g["time"] <= ts)].sort_values("time")

        if len(g) < 2:
            areas.append(np.nan)
            continue

        areas.append(float(trapezoid_fn(g["value"].to_numpy(), g["time"].to_numpy())))

    out = tts_df.copy()
    out["area_of_stability"] = areas
    return out


# Features averaged in the "before"/"after"/"across" table below. F0-F3 are
# the same features TTS is computed on; Jitter/Shimmer/HNR are extra voice
# quality measures that don't get their own stabilization point (see
# STABILIZATION_ANCHOR_FEATURE below) but are still averaged within F0's
# before/after/across windows.
STABILIZATION_MEAN_FEATURES = [
    "F0", "F1", "F2", "F3",
    "Jitter (local)", "Shimmer (local)", "Harmonic to Noise Ratio",
]

# F0 stabilizing is the standard definition of "phonation has stabilized"
# (per your call above) -- this is the single per-recording split point
# used to define "before"/"after", rather than computing a separate split
# per feature.
STABILIZATION_ANCHOR_FEATURE = "F0"


def compute_stabilization_condition_means(time_series_df, tts_df,
                                           anchor_feature=STABILIZATION_ANCHOR_FEATURE,
                                           mean_features=STABILIZATION_MEAN_FEATURES):
    """
    Long-format table: one row per (recordId, condition), condition in
    {"before", "after", "across"}, holding the mean of each of
    `mean_features` within that time window:
      - "before":  time in [t0_time, ts_time) of `anchor_feature`
      - "after":   time in [ts_time, end of recording]
      - "across":  every frame in the recording, regardless of stabilization

    If `anchor_feature` never stabilized for a recording (ts_time is NaN),
    "before"/"after" are NaN for that recording, but "across" is still
    computed since it doesn't depend on ts_time.
    """
    anchor = tts_df[tts_df["feature"] == anchor_feature][
        ["healthCode", "recordId", "t0_time", "ts_time", "time_to_stabilization_s"]
    ].rename(columns={"time_to_stabilization_s": "tts"})

    n_stabilized = anchor["ts_time"].notna().sum()
    print(f"[stabilization_means] anchor feature = {anchor_feature}, "
          f"{n_stabilized}/{len(anchor)} recordings stabilized")

    results = []
    for _, row in anchor.iterrows():
        hc, rid, t0, ts, tts = row["healthCode"], row["recordId"], row["t0_time"], row["ts_time"], row["tts"]
        g = time_series_df[time_series_df["recordId"] == rid].sort_values("time")

        across_means = g[mean_features].mean()

        if pd.notna(ts):
            before_means = g.loc[g["time"] < ts, mean_features].mean()
            after_means = g.loc[g["time"] >= ts, mean_features].mean()
        else:
            dprint(f"[stabilization_means] [{rid}] {anchor_feature} never stabilized -- "
                   f"before/after are NaN, across is still computed")
            before_means = pd.Series({f: np.nan for f in mean_features})
            after_means = pd.Series({f: np.nan for f in mean_features})

        for condition, means in [("before", before_means), ("after", after_means), ("across", across_means)]:
            row_out = {"recordId": rid, "healthCode": hc, "tts": tts, "condition": condition}
            row_out.update(means.to_dict())
            results.append(row_out)

    return pd.DataFrame(results)
```
